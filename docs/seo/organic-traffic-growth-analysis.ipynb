# DataCost organic traffic growth analysis
Reproducible calculations supporting `SEO-TRAFFIC-GROWTH-AUDIT.md`. Search Console values were transcribed from the accessible `sc-domain:datacost.co.za` property on 2026-08-13. Latest reporting date: 2026-08-11.

```python
import pandas as pd

periods = pd.DataFrame([
    {'window': 'Latest 7 days', 'clicks': 439, 'impressions': 107767, 'ctr_pct': 0.4, 'position': 7.7},
    {'window': 'Latest 28 days', 'clicks': 2020, 'impressions': 463000, 'ctr_pct': 0.4, 'position': 7.8},
    {'window': 'Previous 28 days', 'clicks': 1720, 'impressions': 347000, 'ctr_pct': 0.5, 'position': 7.9},
    {'window': 'Latest 3 months', 'clicks': 4444, 'impressions': 957198, 'ctr_pct': 0.5, 'position': 8.0},
    {'window': 'Previous 3 months', 'clicks': 242, 'impressions': 44610, 'ctr_pct': 0.5, 'position': 9.9},
])
periods
```

```python
latest_28, previous_28 = periods.iloc[1], periods.iloc[2]
comparison_28d = pd.Series({
    'click_change_pct': (latest_28.clicks / previous_28.clicks - 1) * 100,
    'impression_change_pct': (latest_28.impressions / previous_28.impressions - 1) * 100,
    'ctr_change_percentage_points': latest_28.ctr_pct - previous_28.ctr_pct,
    'position_change': latest_28.position - previous_28.position,
}).round(2)
comparison_28d
```

```python
segments = pd.DataFrame([
    {'segment': 'South Africa', 'clicks': 1803, 'impressions': 422594},
    {'segment': 'Mobile', 'clicks': 1698, 'impressions': 392350},
])
segments['click_share_pct'] = (segments.clicks / latest_28.clicks * 100).round(1)
segments['impression_share_pct'] = (segments.impressions / latest_28.impressions * 100).round(1)
segments
```

## Interpretation guardrails
The three-month comparison is launch-skewed and is directional only. The latest-versus-previous 28-day comparison drives prioritisation. Rounded Search Console card values mean calculated percentage changes are approximate. GA4 was not used because the accessible Analytics account exposed FreeHub, not a confirmed DataCost property.
